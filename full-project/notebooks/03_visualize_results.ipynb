# Visualization Notebook
Plot training metrics and visualize model predictions.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```
# 1. Plot Training Metrics from CSV
csv_path = "../logs/mobilenet_v3_large_aug_metrics.csv"
if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
    df_clean = df.groupby('epoch').apply(lambda x: x.bfill().ffill().iloc[0]).reset_index(drop=True)
    df_clean['epoch'] = df_clean.index
    
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))
    
    # Loss
    ax1.plot(df_clean['epoch'], df_clean['train_loss'], label='Train Loss')
    ax1.plot(df_clean['epoch'], df_clean['val_loss'], label='Val Loss')
    ax1.set_title('Total Loss')
    ax1.legend()
    
    # Age MAE
    ax2.plot(df_clean['epoch'], df_clean['train_age_mae'], label='Train Age MAE')
    ax2.plot(df_clean['epoch'], df_clean['val_age_mae'], label='Val Age MAE')
    ax2.set_title('Age MAE')
    ax2.legend()
    
    # Gender Accuracy
    ax3.plot(df_clean['epoch'], df_clean['train_gender_acc'], label='Train Gender Acc')
    ax3.plot(df_clean['epoch'], df_clean['val_gender_acc'], label='Val Gender Acc')
    ax3.set_title('Gender Accuracy')
    ax3.legend()
    
    plt.tight_layout()
    plt.show()
else:
    print(f"Metrics CSV not found at {csv_path}")
```

```
# 2. Visualize Predictions on a few images
import sys
sys.path.insert(0, os.path.abspath('..'))
from src.age_gender_model.inference import AgeGenderPredictor
from PIL import Image

try:
    predictor = AgeGenderPredictor(checkpoint_path="../model_store/ag_classifier_main_mobilenet_v3_large_aug_epoch24_loss1.5.pth")
    img = Image.open("../test-images/face14.jpg").convert("RGB")
    res = predictor.predict(img)
    
    plt.imshow(img)
    plt.title(f"Age: {res['age']:.1f} | Gender: {res['gender_label']}")
    plt.axis('off')
    plt.show()
except Exception as e:
    print("Could not run visual predictions. Check paths.", e)
```
